#### Lab Exercise 0: "Tidy" the textual corpus

Complete the function below to tidy this week's corpus. In the end, there should be one row per sentence.

1. Use [the `.explode()` method](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.explode.html) to "unnest" or "explode" the `"tokens"` column.
2. Use [the `.reset_index()` method](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.reset_index.html) to create a unique row index, and move the non-unique index to a new column called `"novel_index"`.
3. Use [the `.groupby()` method](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.groupby.html) and [the `.cumcount()` method](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.api.typing.DataFrameGroupBy.cumcount.html) to create a `"sentence_index"` column.

As with other `import_` functions this term, the function has *no input*. It is for importing a fixed dataset into a research document, so it doesn't need to vary with multiple inputs.

```python
def import_sentiment_analysis_corpus(verbose=2):
    """Import the sentiment analysis corpus from the "data" directory,
    tokenising by sentence.
    
    NB: This function has no input. The "data" directory is hard-coded
    as the location of the corpus."""

    # Import the corpus in the "data" folder, tokenising by sentence
    # Problem: the `tokens` column is a `list`-column, which isn't very useful...
    corpus = import_corpus(corpus_dir="data", tokenise_by="sent", verbose=verbose)

    # Explode the "tokens" column
    ### BEGIN SOLUTION
    tidy_corpus = corpus.explode("tokens")
    ### END SOLUTION
    
    # Reset the index, storing the old row index in a new column called "novel_index"
    ### BEGIN SOLUTION
    tidy_corpus = tidy_corpus.reset_index(names="novel_index")
    ### END SOLUTION

    # Create the "sentence_index" column using .groupby() and .cumcount()
    ### BEGIN SOLUTION
    tidy_corpus = tidy_corpus.assign(sentence_index = tidy_corpus.groupby("title").cumcount())
    ### END SOLUTION

    # Return!
    return tidy_corpus
```
